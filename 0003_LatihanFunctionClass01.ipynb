### Python Classes and the `__init__` method

In Python, a **class** is a blueprint for creating objects (a particular data structure), providing initial values for state (member variables or attributes), and implementations of behavior (member functions or methods).

The `__init__` method is a special method, also known as a **constructor**, that gets automatically called whenever a new object of the class is created. It is used to initialize the attributes of the object. The `self` parameter refers to the instance of the class itself, and it must be the first parameter of any instance method, including `__init__`.

```python
class MyClass:
    def __init__(self, name, age):
        self.name = name  # Initialize the 'name' attribute
        self.age = age    # Initialize the 'age' attribute
        print(f"A new object named {self.name} has been created!")

    def introduce(self):
        return f"Hello, my name is {self.name} and I am {self.age} years old."

# Create an instance of MyClass
person1 = MyClass("Alice", 30)

# Access attributes and call methods
print(person1.introduce())
```

### Contoh Pemanggilan Langsung pada Objek Sementara

Jika Anda hanya ingin menampilkan informasi dari objek yang baru dibuat dan tidak berencana menggunakan objek tersebut lagi setelah itu, Anda bisa melakukannya dalam satu ekspresi seperti ini:

```python
# Membuat objek Student dan langsung memanggil display_info()
print("\n--- Contoh Pemanggilan Langsung ---")
Student("Cici", 9, "4B", 102).display_info()
```

Dalam contoh di atas, objek `Student("Cici", 9, "4B", 102)` dibuat, metode `display_info()` langsung dipanggil padanya, tetapi objek itu sendiri tidak disimpan dalam variabel apa pun. Jadi, Anda tidak bisa lagi mengakses objek 'Cici' ini setelah baris tersebut selesai.

**Kesimpulan:**
Sebagian besar waktu, Anda akan ingin menyimpan objek dalam variabel (seperti `student_a`) karena lebih mudah untuk dikelola dan digunakan kembali di bagian kode lain.

In this example:
- We define a class `MyClass`.
- The `__init__` method takes `self`, `name`, and `age` as arguments. It initializes `self.name` and `self.age` with the values passed during object creation.
- When we create `person1 = MyClass("Alice", 30)`, the `__init__` method is called, and `name` becomes "Alice" and `age` becomes 30 for this specific object.
- The `introduce` method is a regular method that uses the object's `name` and `age` attributes.
