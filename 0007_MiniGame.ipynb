```python
import random as r
from IPython.display import clear_output

class Engine:
  def __init__(self):
    self.num3 = 0
    self.check = ""
    self.num4 = 0
    self.skor = 0

  def clear(self):
    pass

  def generator(self):
    self.num3 = r.randint(1,10)
    self.num4 = r.randint(1,10)
    self.check = r.choice(["+","-","*","/"])
    self.loop()

  def loop(self):
    while(True):
      clear_output()
      print(f" {self.num3} {self.check} {self.num4}")
      user = input(f"masukan hasil: ")
      if(user.replace('.','',1).isdigit()): # handles simple integers and floats
        user=float(user)
        if(self.check == "+"):
          if(user == (self.num3+self.num4)):
            self.skor += 10
            print(f"Jawaban Benar. skor {self.skor}")
          else:
            print(f"jawaban salah {self.skor}")
            break

        if(self.check == "-"):
          if(user == (self.num3-self.num4)):
            self.skor += 10
            print(f"Jawaban Benar. skor {self.skor}")
          else:
            print("salah")
            break

        if(self.check == "*"):
          if(user == (self.num3*self.num4)):
            self.skor += 20
            print(f"Jawaban Benar. skor {self.skor}")
          else:
            print("salah")
            break

        if(self.check == "/"):
          if(user == (self.num3/self.num4)):
            self.skor += 20
            print(f"Jawaban Benar. skor {self.skor}")
          else:
            print("salah")
            break
      else:
        continue
      self.generator()


if __name__ == '__main__':
  game = Engine()
  game.generator()
```
