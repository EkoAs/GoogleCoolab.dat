```python
class Tank:
  def __init__(self, name="", armor="", engine="", cannon="", hp="", damage=""):
    self.name = name
    self.armor = armor
    self.engine = engine
    self.cannon = cannon
    self.hp = hp
    self.damage = damage


  def select(self):
    options = {
        "name": ["M41", "T-90", "Leopard-2", "Changeler-2", "Type 71", "T-72"],
        "armor": ["Light 100mm","Medium 200mm","Heavy 350mm","Heat 250mm", "Composite 170mm"],
        "engine": ["V8 300hp","V8 670hp TwinTurbo","V12 550hp","V12 800hp Turbo", "V16 890hp", "X16 Charge 950hp"],
        "cannon":["80mm","120mm","150mm","180mm","130mm + 12mm", "120mm + 15mm + 12mm"]
    }



    for catego, items  in options.items():

      for index, item in enumerate(items, start=1):
        print(f"{index}. {item}")


      while(True):
        try:
          opsi = int(input(f"\nSelect {catego} (1-{len(item)}): "))
          if 1 <= opsi <= len(items):
            itemSelect = items[opsi-1]

            setattr(self, catego, itemSelect)
            print(f"-> {catego.capitalize()} terpilih: {itemSelect}")
            break
          else:
            print("Pilihan tidak valid. Silakan coba lagi.")
        except ValueError:
          print("Input harus berupa angka.")


tank_ku = Tank()
tank_ku.select()
```
