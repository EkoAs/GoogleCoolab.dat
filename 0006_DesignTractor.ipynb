```python
from IPython.display import clear_output

class Tractor:
  def __init__(self, name="", model="", color="", engine=0, whell="", work="", cost=""):
    self.name = name
    self.model = model
    self.color = color
    self.engine = engine
    self.whell = whell
    self.work = work
    self.cost = cost
    self.shop = {
        "Fendt": 500000,
        "Jhon Degree": 430000,
        "Deutz-Fahr": 350000,
        "Valtra": 550000,
        "New Holland": 190000,
        "Claas": 150000
    }
    self.all_tractor_data = {
      "Fendt": {
          "Vario 200 Series": {"price": 150000},
          "Vario 700 Series": {"price": 250000},
          "Vario e100": {"price": 48000}
      },
      "Jhon Degree": {
          "5R Series": {"price": 180000},
          "7R Series": {"price": 280000},
          "6120M": {"price": 380000}
      },
      "Deutz-Fahr": {
          "5 Series": {"price": 140000},
          "6 Series": {"price": 220000},
          "9 Series": {"price": 350000}
      },
      "Valtra": {
          "A Series": {"price": 130000},
          "T Series": {"price": 200000},
          "S Series": {"price": 300000}
      },
      "New Holland": {
          "T5 Series": {"price": 160000},
          "T7 Series": {"price": 240000},
          "T8 Series": {"price": 330000}
      },
      "Claas": {
          "Arion 400 Series": {"price": 170000},
          "Axion 800 Series": {"price": 270000},
          "Xerion 5000 Series": {"price": 420000}
      }
    }

  def user(self):
    print(f"You must add over 1000000")
    money = input(f"Enter your money: ")
    if money.isdigit():
      money = int(money)
      if(money >= 1000000):
        self.cost = money
      else:
        print("Input must over 1000000.")
        return self.user()
    else:
      print("Invalid input. Please enter a valid number.")
      return self.user()
    clear_output()
    self.logic()

  def logic(self):
    num=0
    for name, data, in self.shop.items():
      num+=1
      print(f"\n Options -{num} Brand {name} price {data}")
      user = input(f"Pilih Shop (1 untuk Ya, 0 untuk Lewati): ")
      if(user.isdigit()):
        user=int(user)
      else:
        print("Invalid input. Please enter a valid number.")
        return self.logic()

      if(user == 1):
        clear_output()
        self.name = name
        self.cost = self.cost - data
        print(f"\n Best choice {name}")
        print(f"Your Money: {self.cost}")
        self.all_tractor()
        break
      clear_output()

  def all_tractor(self):
    print(f"\n\n--- PILIH MODEL UNTUK {self.name} ---")
    print(f"Type 1 to yes, 0 to skip\n")

    if self.name in self.all_tractor_data:
      models = self.all_tractor_data[self.name]
      num = 0
      for model_name, detail in models.items():
        num += 1
        price = detail['price']
        print(f"\n Option -{num} Model {model_name} Price {price}")
        user_choice = input(f"Pilih Model ini? (1): ")

        if user_choice.isdigit():
          user_choice = int(user_choice)
        else:
          continue

        if user_choice == 1:
          clear_output()
          self.model = model_name
          self.cost = self.cost - price
          print(f"\n Model Selected: {self.model}")
          print(f"Remaining Money: {self.cost}")
          print(f"Summary: {self.name} {self.model} is ready!")
          break
        clear_output()
    else:
      print("Data brand tidak ditemukan.")

if __name__ == "__main__":
  Tractor().user()
```
