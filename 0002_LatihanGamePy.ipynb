```python
import random as rd
import os

daftar = []
kosakata = ""

def clear():
    if os.name == 'nt':
        _ = os.system('cls')
    else:
        _ = os.system('clear')

def kata(kosakata,daftar):
    kosakata=input(f"masukan kosakata")
    total = input("masukan angka jumlah kata: ")
    if total.isdigit():
        total = int(total)
    else:
        print("masukan angka")
        return kata(daftar)

    # Menggunakan 'total' sesuai hasil input, bukan 'num'
    for i in range(total):
        user = input(f"masukan Kata ke {i+1}: ")
        daftar.append(user)

    clear()

# Menyesuaikan parameter agar menerima 'kosakata' dan 'daftar'
def tebak(kosakata, daftar):
    print(f"ciri2 nya adalah {kosakata}")

    percobaan = len(daftar)
    while percobaan > 0:
        user_pil = input("\nmasukan kata: ")
        if user_pil in daftar:
            print(f"selamat, pilihan {user_pil} anda ada!")
            break
        else:
            print("Pilihan anda salah")
        percobaan -= 1

kata(kosakata, daftar)

if daftar:
    kosakata = rd.choice(daftar)
    tebak(kosakata, daftar)
```
