```python
class Tank:
  def __init__(self, name="", armor="", engine="", cannon="", hp="", damage="", mobility=0):
    self.name = name
    self.armor = armor
    self.engine = engine
    self.cannon = cannon
    self.hp = hp
    self.damage = damage
    self.mobility = mobility
  armor = 0

  def select(self):
    options = {
        "name": ["M41", "T-90", "Leopard-2", "Changeler-2", "Type 71", "T-72"],
        "armor": ["Light 100mm","Medium 200mm","Heavy 350mm","Heat 250mm", "Composite 170mm"],
        "engine": ["V8 300hp","V8 670hp TwinTurbo","V12 550hp","V12 800hp Turbo", "V16 890hp", "X16 Charge 950hp"],
        "cannon":["80mm","120mm","150mm","180mm","130mm + 12mm", "120mm + 15mm + 12mm"]
    }



    for catego, items  in options.items():

      for index, item in enumerate(items, start=1):
        print(f"{index}. {item}")


      while(True):
        try:
          opsi = int(input(f"\nSelect {catego} (1-{len(item)}): "))
          if 1 <= opsi <= len(items):
            itemSelect = items[opsi-1]

            setattr(self, catego, itemSelect)
            print(f"-> {catego.capitalize()} terpilih: {itemSelect}")
            break
          else:
            print("Pilihan tidak valid. Silakan coba lagi.")
        except ValueError:
          print("Input harus berupa angka.")

  def logic(self):
    self.hp = 200
    self.damage=0
    mobility = 0
    global armor
    armor = 0


    if self.armor == "Light 100mm":
      self.hp += 100
      self.damage += 10
      self.mobility = 10
      armor += 10
    if self.armor == "Medium 200mm":
      self.hp += 230
      self.damage += 15
      self.mobility = 35
      armor +=20
    if self.armor == "Heavy 350mm":
      self.hp += 300
      self.damage += 20
      self.mobility =60
      armor += 35
    if self.armor == "Heat 250mm":
      self.hp += 150
      self.damage += 50
      self.mobility = 24
      armor +=25
    if self.armor == "Composite 170mm":
      self.hp += 200
      self.damage += 30
      self.mobility = 35
      armor += 17


    if self.engine == "V8 300hp":
      self.hp += 80
      self.damage += 0
      self.mobility += 150
      armor += 15
    if self.engine == "V8 670hp TwinTurbo":
      self.hp += 80
      self.damage += 0
      self.mobility += 330
      armor += 37
    if self.engine == "V12 550hp":
      self.hp += 120
      self.damage += 10
      self.mobility += 250
      armor += 25
    if self.engine == "V12 800hp Turbo":
      self.hp +=130
      self.damage += 15
      self.mobility += 400
      armor += 40
    if self.engine == "V16 890hp":
      self.hp +=140
      self.damage += 18
      self.mobility += 450
      armor += 45
    if self.engine == "X16 Charge 950hp":
      self.hp +=140
      self.damage += 25
      self.mobility += 620
      armor += 50

    if self.cannon == "80mm":
      self.hp += 0
      self.damage += 40
      self.mobility += 8
      armor += 8
    if self.cannon == "120mm":
      self.hp += 12
      self.damage += 60
      self.mobility += 20
      armor += 12
    if self.cannon == "150mm":
      self.hp += 20
      self.damage += 80
      self.mobility += 25
      armor += 15
    if self.cannon == "180mm":
      self.hp += 30
      self.damage += 100
      self.mobility += 35
      armor += 20
    if self.cannon == "130mm + 12mm":
      self.hp += 15
      self.damage += (80+12)
      self.mobility += 15
      armor += 15
    if self.cannon == "120mm + 15mm + 12mm":
      self.hp += 10
      self.damage += (60+15+12)
      self.mobility += 18
      armor += 12

  def display(self):
    print("\n\n\n")

    print(f"Name: {self.name}")
    print(f"Armor: {self.armor}")
    print(f"Engine: {self.engine}")
    print(f"Cannon: {self.cannon}")

    print(f"HP: {self.hp}")
    print(f"Damage: {self.damage}")

    print("\n\n\n")
    print(f"Mobility: {self.mobility}")
    print(f"Armor: {armor}")
    mobility_1 = float(self.mobility / armor)
    print(f"Mobility: {mobility_1:.2f}")

tank_ku = Tank()
tank_ku.select()
tank_ku.logic()
tank_ku.display()
```
