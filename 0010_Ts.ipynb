```python
import random as rd
from google.colab import output

def l():
  poin=0
  while(True):
    n = rd.randint(1,10)
    n2 = rd.randint(1,10)
    output.clear()
    print(f" Poin: {poin}")

    user = input(f"{n} x {n2} : ")
    if(user.isdigit()):
      user=int(user)
      if(user == (n * n2)):
        poin+=10
        print(f"+10 poin")
      else:
        print(f"result {n * n2}")
        print(f"uncorret, scor: {poin}")
        break
    else:
      print("Input not int.")
      break
l()
```
