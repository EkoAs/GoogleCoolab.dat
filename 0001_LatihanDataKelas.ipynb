```python
data = []
datanim = []

def induk():
  def name():
    nama = input(f"masukan nama kamu: ")
    if nama.strip():
      print(f"halo {nama}")
      return nama
    else:
      print("masukan nama anda kembali!")
      return name()

  def nik():
    print(f"sambutan sudah selesai, sekarang isi nim: ")
    nim_in = input("masukan 5 digit nim: ")
    if nim_in.isdigit():
      nim_val = int(nim_in)
      if nim_val in datanim:
        print(f"NIM {nim_val} sudah terdaftar!")
      else:
        datanim.append(nim_val)
      return nim_val
    else:
      print(f"nim Bukan digit!")
      return nik()

  def clas():
    isi = input("masukan kelas: ")
    if not isi.isdigit():
      print("Harus angka!")
      return clas()
    kelas = int(isi)
    if kelas in data:
      print(f"anda sudah ada di kelas {kelas}")
      return clas()
    else:
      data.append(kelas)
      print(f'selamat datang di kelas {kelas}')
    return data

  user_name = name() # kenapa harus jadi gini user_name = name()
  user_nim = nik() # user_nim = nik()
  user_data = clas() # user_data = clas()
  return user_name, user_nim, user_data # return user_name, user_nim, user_data

while True:
  hasil = induk()
  # ini adalah data tuple dari diatas juga karena funct in functi
  nama_user = hasil[0] # nama = hasil[0]
  current_data = hasil[2] # nim = hasil[2] , dan kelas = hasil[3]

  print(f"--- Laporan: {nama_user} | Total Kelas Terdaftar: {len(data)} ---")
  print(f"Daftar Semua Kelas: {data}")

  if len(data) >= 3:
    print(f"Batas kelas tercapai!")
    break
  else:
    tanya = input("Tambah data lagi? (y/n): ")
    if tanya.lower() != 'y':
      break
```
